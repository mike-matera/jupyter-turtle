# Test the Statistics on the Simple API

I use this for my CIS-15 class. In that class I use the nb_unittest project to give students feedback. The simple API keeps track of statistics to help me.

```python
%load_ext jupyter_turtle
import jupyter_turtle as tu
```

```python
tu.move(100)
assert tu._turtle.stats["moves"] == 1
assert tu._turtle.stats["strokes"] == 1
assert tu._turtle.stats["distance"] == 100
```

```python
tu.goto(100, 100)
assert tu._turtle.stats["moves"] == 1
assert tu._turtle.stats["strokes"] == 1
assert round(tu._turtle.stats["distance"]) == round(141.421356237)
```

```python
tu.pen_up()
tu.move(100)
assert tu._turtle.stats["moves"] == 1
assert tu._turtle.stats["strokes"] == 0
assert tu._turtle.stats["distance"] == 0
```

```python
tu.pen_up()
tu.move(100)
tu.pen_down()
tu.move(-100)
assert tu._turtle.stats["moves"] == 2
assert tu._turtle.stats["strokes"] == 1
assert tu._turtle.stats["distance"] == 100
```

```python
for _ in range(4):
    tu.move(100)
    tu.turn(90)
assert tu._turtle.stats["moves"] == 4
assert tu._turtle.stats["strokes"] == 1
assert tu._turtle.stats["distance"] == 400
assert tu._turtle.stats["turns"] == 4
```

```python
tu.goto(0,100)
tu.goto(100,100)
tu.goto(100,0)
tu.goto(0,0)
assert tu._turtle.stats["moves"] == 4
assert tu._turtle.stats["strokes"] == 1
assert tu._turtle.stats["distance"] == 400
assert tu._turtle.stats["turns"] == 0
assert tu._turtle.stats["jumps"] == 4
```

```python
tu.set_heading(0)
tu.move(100)
tu.set_heading(90)
tu.move(100)
tu.set_heading(180)
tu.move(100)
tu.set_heading(270)
tu.move(100)
assert tu._turtle.stats["moves"] == 4
assert tu._turtle.stats["strokes"] == 1
assert tu._turtle.stats["distance"] == 400
assert tu._turtle.stats["turns"] == 4
```

```python
tu.pen_up()
tu.write("Python")
tu.turn(30)
tu.move(-30)
tu.write("is")
tu.turn(30)
tu.move(-30)
tu.write("fun")
tu.move(-30)
assert tu._turtle.stats["moves"] == 3
assert tu._turtle.stats["strokes"] == 0
assert tu._turtle.stats["distance"] == 0
assert tu._turtle.stats["turns"] == 2
assert tu._turtle.stats["words"] == ["Python", "is", "fun"]
```
